```python
#I HAVE IMPORTED THE WINE QUALITY DATASET FROM UCI MACHINE LEARNING REPOSITORY WHICH HAS TWO CSV FILES:


# Import necessary libraries. I import pandas for data manipulation and analysis.
import pandas as pd

# Load the datasets. There are 2 csv files .
red_wine = pd.read_csv('winequality-red.csv', sep=';')
white_wine = pd.read_csv('winequality-white.csv',sep=';')

# Display the first few rows of each dataset
print("Red Wine Dataset:")
print(red_wine.head())

print("\nWhite Wine Dataset:")
print(white_wine.head())

# Check for missing values in both datasets
print("\nMissing values in Red Wine Dataset:")
print(red_wine.isnull().sum())

print("\nMissing values in White Wine Dataset:")
print(white_wine.isnull().sum())

# Clean the datasets (if necessary)
# For example, drop rows with missing values
red_wine_cleaned = red_wine.dropna()
white_wine_cleaned = white_wine.dropna()

# Display the cleaned datasets' info
print("\nRed Wine Dataset after cleaning:")
print(red_wine_cleaned.info())

print("\nWhite Wine Dataset after cleaning:")
print(white_wine_cleaned.info())


```

```python
# Compute basic statistics (e.g., mean, median, standard deviation) for numerical columns in both datasets
print("\nBasic Statistics for Red Wine Dataset:")
print(red_wine_cleaned.describe())

print("\nBasic Statistics for White Wine Dataset:")
print(white_wine_cleaned.describe())

# Perform groupings on a categorical column and compute the mean of a numerical column
# In this dataset, the 'quality' column is categorical, and we can group by it
print("\nAverage values for each quality level in Red Wine Dataset:")
print(red_wine_cleaned.groupby('quality').mean())

print("\nAverage values for each quality level in White Wine Dataset:")
print(white_wine_cleaned.groupby('quality').mean())

# Identify patterns or interesting findings
# For example, check if higher quality wines have higher alcohol content
print("\nAverage alcohol content by quality in Red Wine Dataset:")
print(red_wine_cleaned.groupby('quality')['alcohol'].mean())

print("\nAverage alcohol content by quality in White Wine Dataset:")
print(white_wine_cleaned.groupby('quality')['alcohol'].mean())

# Check for correlations between different features
print("\nAverage residual surgar by quality in Red Wine dataset:")
print(red_wine_cleaned.groupby('quality')['residual sugar'].mean())
```

```python
# Import necessary libraries for visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Line Chart: Example of a trend (if time-series data is available, otherwise skip)
# Assuming 'quality' is sorted, we can visualize trends in alcohol content
plt.figure(figsize=(10, 6))
sns.lineplot(data=red_wine_cleaned.groupby('quality')['alcohol'].mean(), marker='o', label='Red Wine')
sns.lineplot(data=white_wine_cleaned.groupby('quality')['alcohol'].mean(), marker='o', label='White Wine')
plt.title('Average Alcohol Content by Quality')
plt.xlabel('Quality')
plt.ylabel('Average Alcohol Content')
plt.legend()
plt.show()

# Bar Chart: Comparison of average residual sugar across wine qualities
plt.figure(figsize=(10, 6))
sns.barplot(x=red_wine_cleaned['quality'], y=red_wine_cleaned['residual sugar'], ci=None, color='red', label='Red Wine')
sns.barplot(x=white_wine_cleaned['quality'], y=white_wine_cleaned['residual sugar'], ci=None, color='blue', label='White Wine')
plt.title('Average Residual Sugar by Quality')
plt.xlabel('Quality')
plt.ylabel('Residual Sugar')
plt.legend()
plt.show()

# Histogram: Distribution of alcohol content in red and white wines
plt.figure(figsize=(10, 6))
sns.histplot(red_wine_cleaned['alcohol'], bins=20, kde=True, color='red', label='Red Wine', alpha=0.6)
sns.histplot(white_wine_cleaned['alcohol'], bins=20, kde=True, color='blue', label='White Wine', alpha=0.6)
plt.title('Distribution of Alcohol Content')
plt.xlabel('Alcohol Content')
plt.ylabel('Frequency')
plt.legend()
plt.show()

# Scatter Plot: Relationship between alcohol content and residual sugar
plt.figure(figsize=(10, 6))
sns.scatterplot(x=red_wine_cleaned['alcohol'], y=red_wine_cleaned['residual sugar'], color='red', label='Red Wine')
sns.scatterplot(x=white_wine_cleaned['alcohol'], y=white_wine_cleaned['residual sugar'], color='blue', label='White Wine')
plt.title('Alcohol Content vs Residual Sugar')
plt.xlabel('Alcohol Content')
plt.ylabel('Residual Sugar')
plt.legend()
plt.show()
```
